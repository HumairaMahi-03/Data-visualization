```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

# Histogram


```python
df['Age'].fillna(df['Age'].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Age'].hist(bins=20, edgecolor='black',figsize=(10,5))
```

```python
df['Fare'].hist(bins = 20,edgecolor= 'black',figsize=(10,5))
```

```python
df['Fare'].mean()
```

```python
df['Fare'].describe()
```

```python
df.sample(5)
```

```python
# Frequency polygon for Age
data = df['Age']
counts, bins = np.histogram(data, bins=10)

# Midpoints
midpoints = (bins[:-1] + bins[1:]) / 2
plt.figure(figsize=(14,4))
plt.hist(data, bins = 10 ,alpha = 0.5)
plt.plot(midpoints, counts , marker='o')
plt.show()
```

```python
data2 = df['Fare']
count2 ,bins2 = np.histogram(data2,bins =10)
midpoints2 = (bins2[:-1] + bins2[1:] )/2

plt.figure(figsize=(14,4))
plt.hist(data2,bins=10,alpha=0.5)
plt.plot(midpoints2,count2,marker='o')
plt.show()
```

```python
# fruency curve for age
sns.kdeplot(df['Age'],shade=True)
plt.title('Age Frequency Curve')
plt.xlabel('Age')
plt.ylabel('Density')
plt.show()
```

```python
# fruency curve for fare
sns.kdeplot(df['Fare'],shade=True)
plt.title('Fare Frequency Curve')
plt.xlabel('Fare')
plt.ylabel('Density')
plt.show()
```
